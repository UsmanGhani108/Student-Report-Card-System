# Student Report Card System

```python
std_name = input("Enter your name : ")
english = int(input("Enter English Marks : "))
math = int(input("Enter Math Marks : "))
urdu = int(input("Enter Urdu Marks : "))
chemistry = int(input("Enter Chemistry Marks : "))
physics = int(input("Enter Physics Marks : "))
std_marks = [english , math , urdu , chemistry , physics]
std_data = {}
std_data ["Name"] = std_name
std_data ["English Marks"] = english
std_data ["Maths Marks"] = math
std_data ["Chemistry Marks"] = chemistry
std_data ["Physics Marks"] = physics
std_data["Urdu Marks"] = urdu
total_marks = 0
for i in std_marks:
    total_marks +=  i
percentage = round(total_marks / 500 * 100)
grade = None
if percentage >= 90 and percentage <= 100:
    grade = "A+"
elif percentage >= 80 and percentage <= 89:
    grade = "A"
elif percentage >= 70 and percentage <= 79:
    grade = "B"
elif percentage >= 60 and percentage <= 69:
    grade = "C"
elif percentage >= 50 and percentage <= 59:
    grade = "D"
else:
    grade = "F"
if grade != "F":
    result = "Pass"
else:
    result = "Fail"
std_data["total Marks"] = total_marks
std_data["Percentage"] = percentage
std_data["Grade"] = grade
std_data["Result"] = result
print(std_data)
```
